1. Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

2. Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
import numpy as np

# 3x4 array of zeros
zeros_arr = np.zeros((3, 4))
print("3x4 array of zeros:")
print(zeros_arr)
print()

# 3x4 array of ones
ones_arr = np.ones((3, 4))
print("3x4 array of ones:")
print(ones_arr)
print()

# 4x4 identity matrix
identity_arr = np.eye(4)
print("4x4 identity matrix:")
print(identity_arr)
print()

# 8 evenly spaced numbers between 0 and 1
linspace_arr = np.linspace(0, 1, 8)
print("8 evenly spaced numbers between 0 and 1:")
print(linspace_arr)
```

3. Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
import numpy as np

# Create a 1D array of numbers from 1 to 24
arr_1d = np.arange(1, 25)
print("1D array:")
print(arr_1d)
print("Shape:", arr_1d.shape)
print()

# Reshape into a 2D array of shape (4, 6)
arr_2d = arr_1d.reshape(4, 6)
print("2D array (4, 6):")
print(arr_2d)
print("Shape:", arr_2d.shape)
print()

# Reshape into a 3D array of shape (2, 3, 4)
arr_3d = arr_1d.reshape(2, 3, 4)
print("3D array (2, 3, 4):")
print(arr_3d)
print("Shape:", arr_3d.shape)
```

4. Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
import numpy as np

arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

# First 3 elements
print("First 3 elements:", arr[:3])

# Last 3 elements
print("Last 3 elements:", arr[-3:])

# Every alternate element (starting from index 0)
print("Every alternate element:", arr[::2])

# Array in reverse order
print("Reversed array:", arr[::-1])
```

5. Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
import numpy as np

# Create a 5x5 array with values 1 to 25
arr = np.arange(1, 26).reshape(5, 5)
print("Original 5x5 array:")
print(arr)
print()

# (a) Second row (index 1)
second_row = arr[1]
print("(a) Second row:", second_row)
print()

# (b) Third column (index 2)
third_col = arr[:, 2]
print("(b) Third column:", third_col)
print()

# (c) 3x3 sub-matrix from the center
center = arr[1:4, 1:4]
print("(c) Center 3x3 sub-matrix:")
print(center)
```

6. Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

# Element-wise operations
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a ** 2 =", a ** 2)
print()

# Dot product
print("Dot product of a and b =", np.dot(a, b))
```

7. Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
import numpy as np

arr = np.array([[4, 8, 2],
                [9, 1, 7],
                [5, 6, 3]])

# Whole-array statistics
print("Array:")
print(arr)
print()
print("Sum of whole array:          ", arr.sum())
print("Mean of whole array:         ", arr.mean())
print("Min of whole array:          ", arr.min())
print("Max of whole array:          ", arr.max())
print("Standard deviation:          ", arr.std())
print()

# Along rows (axis=1) → one value per row
print("Sum along rows (axis=1):     ", arr.sum(axis=1))
print("Mean along rows (axis=1):    ", arr.mean(axis=1))
print()

# Along columns (axis=0) → one value per column
print("Sum along columns (axis=0):  ", arr.sum(axis=0))
print("Mean along columns (axis=0): ", arr.mean(axis=0))
```

8. Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
import numpy as np

# Create an array of 20 random integers between 1 and 100
rng = np.random.default_rng(seed=42)   # seed for reproducibility
arr = rng.integers(1, 101, size=20)
print("Array:", arr)
print()

# (a) All elements greater than 50
greater_than_50 = arr[arr > 50]
print("(a) Elements > 50:", greater_than_50)
print()

# (b) All even elements
even_elements = arr[arr % 2 == 0]
print("(b) Even elements:", even_elements)
print()

# (c) Count of elements divisible by 5
count_div_by_5 = np.sum(arr % 5 == 0)
print("(c) Count of elements divisible by 5:", count_div_by_5)
```

9. Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
import numpy as np

arr = np.array([5, -3, 8, -1, 0, -7, 4])
print("Original array:", arr)
print()

# Replace all negative values with 0 using np.where
arr_replaced = np.where(arr < 0, 0, arr)
print("After replacing negatives with 0:", arr_replaced)
print()

# Find index of maximum value
max_idx = np.argmax(arr)
print("Index of max value:", max_idx)
print("Max value:", arr[max_idx])
print()

# Find index of minimum value
min_idx = np.argmin(arr)
print("Index of min value:", min_idx)
print("Min value:", arr[min_idx])
```

10. Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
import numpy as np

# Create two 2x3 arrays
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([[7, 8, 9],
              [10, 11, 12]])

print("Array a:")
print(a)
print()
print("Array b:")
print(b)
print()

# Vertical stack (vstack) → shape (4, 3)
v_stack = np.vstack((a, b))
print("Vertical stack (vstack):")
print(v_stack)
print("Shape:", v_stack.shape)
print()

# Horizontal stack (hstack) → shape (2, 6)
h_stack = np.hstack((a, b))
print("Horizontal stack (hstack):")
print(h_stack)
print("Shape:", h_stack.shape)
print()

# Sort along each row and flatten
arr = np.array([[9, 3, 6],
                [2, 8, 1]])
print("Original array:")
print(arr)
print()

sorted_arr = np.sort(arr, axis=1)   # sort along each row
print("Sorted along rows (axis=1):")
print(sorted_arr)
print()

flattened = sorted_arr.flatten()
print("Flattened 1D array:", flattened)
print("Shape:", flattened.shape)
```
