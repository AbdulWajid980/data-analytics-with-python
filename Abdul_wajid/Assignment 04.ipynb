1. Basic Dictionary Creation
Create a dictionary called student with keys: name, age, course, city. Print each value

```python
student = {
    "name": "Ali",
    "age": 20,
    "course": "Python",
    "city": "Lahore"
}
print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("City:", student["city"])
```

2. Adding & Updating Values

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}
car["color"] = "White"
car["year"] = 2022
print(car)
```

3. Loop Through Dictionary

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

4. Count Frequency of Numbers

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1
print(frequency)
```

5. Word Counter

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1
print("\nWord Frequency:")
for word, count in frequency.items():
    print(word, ":", count)
```

6. Student Marks & Average

```python
marks = {"Math": 80, "English": 70, "Physics": 90}
total = sum(marks.values())
average = total / len(marks)
print("Marks:", marks)
print("Total:", total)
print("Average:", average)
```

7.  Highest Scoring Student

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}
highest_score = max(students.values())
for name, score in students.items():
    if score == highest_score:
        print("Highest scoring student:", name)
        print("Score:", score)
```

8.  Merge Dictionaries

```python

dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
merged = dict1.copy()
merged.update(dict2)
print("Merged dictionary:", merged)
```
