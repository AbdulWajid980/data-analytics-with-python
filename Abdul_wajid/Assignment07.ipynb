6. Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd

data = {
    "Name":   ["Ahmed", "Fatima", "Bilal", "Ayesha", "Usman", "Zainab"],
    "Age":    [25, 30, 35, 28, 42, 31],
    "City":   ["Karachi", "Lahore", "Islamabad", "Faisalabad", "Rawalpindi", "Peshawar"],
    "Salary": [55000, 72000, 88000, 61000, 95000, 67000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print()


df["Bonus"] = df["Salary"] * 0.10
print("After adding 'Bonus' column:")
print(df)
print()

df["Total"] = df["Salary"] + df["Bonus"]
print("After adding 'Total' column:")
print(df)
print()


df = df.rename(columns={"City": "Location"})
print("After renaming 'City' → 'Location':")
print(df)
print()

df = df.drop(columns=["Bonus"])
print("After dropping 'Bonus' column:")
print(df)
```

7. Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd
import numpy as np

data = {
    "Name":   ["Ahmed", "Fatima", "Bilal", "Ayesha", "Usman", "Zainab", "Hassan"],
    "Age":    [25, np.nan, 35, 28, np.nan, 31, 29],
    "City":   ["Karachi", "Lahore", np.nan, "Faisalabad", "Rawalpindi", "Peshawar", np.nan],
    "Salary": [55000, 72000, np.nan, 61000, 95000, np.nan, 58000]
}

df = pd.DataFrame(data)
print("Original DataFrame with missing values:")
print(df)
print()

print("Missing values per column:")
print(df.isnull().sum())
print()

# Bonus: total and percentage
print("Total missing values:", df.isnull().sum().sum())
print()
print("Percentage missing per column:")
print((df.isnull().sum() / len(df) * 100).round(2))
print()


numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("After filling numeric NaNs with column mean:")
print(df)
print()

df_clean = df.dropna()

print("After dropping rows with any remaining NaN:")
print(df_clean)
print()
print("Shape before:", df.shape)
print("Shape after: ", df_clean.shape)
```

8. Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
import pandas as pd


data = {
    "Name":   ["Ahmed", "Fatima", "Bilal", "Ayesha", "Usman", "Zainab"],
    "Age":    [25, 30, 35, 28, 42, 31],
    "City":   ["Karachi", "Lahore", "Islamabad", "Faisalabad", "Rawalpindi", "Peshawar"],
    "Salary": [55000, 72000, 88000, 61000, 95000, 67000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print()

# ---------------------------------------------------------------
# 1. Sort by Salary (descending), then by Age (ascending)
# ---------------------------------------------------------------
df_sorted = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted by Salary (desc) then Age (asc):")
print(df_sorted)
print()

# ---------------------------------------------------------------
# 2. Top 3 highest-paid employees using nlargest()
# ---------------------------------------------------------------
top3 = df.nlargest(3, "Salary")
print("Top 3 highest-paid employees:")
print(top3)
```

9. Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
import pandas as pd

data = {
    "Name":   ["Ahmed", "Fatima", "Bilal", "Ayesha", "Usman",
               "Zainab", "Hassan", "Maryam", "Imran", "Sana"],
    "Age":    [25, 30, 35, 28, 42, 31, 29, 38, 45, 27],
    "City":   ["Karachi", "Lahore", "Islamabad", "Karachi", "Lahore",
               "Karachi", "Islamabad", "Lahore", "Karachi", "Islamabad"],
    "Salary": [55000, 72000, 88000, 61000, 95000,
               67000, 58000, 82000, 99000, 63000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print()

print("1. Average salary per City:")
print(df.groupby("City")["Salary"].mean())
print()

print("2. Number of employees per City:")
print(df.groupby("City")["Name"].count())
# or equivalently: df.groupby("City").size()
print()

print("3. Maximum salary per City:")
print(df.groupby("City")["Salary"].max())
print()

print("4. Average, count, and max salary per City using agg():")
result = df.groupby("City")["Salary"].agg(["mean", "count", "max"])
print(result)
print()

print("5. Named aggregation with custom column names:")
result_named = df.groupby("City")["Salary"].agg(
    Avg_Salary="mean",
    Num_Employees="count",
    Max_Salary="max"
)
print(result_named)
print()

print("6. Multiple columns aggregated together:")
result_multi = df.groupby("City").agg({
    "Salary": ["mean", "max", "min"],
    "Age":    ["mean", "count"]
})
print(result_multi)
```

10. Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd


employees = pd.DataFrame({
    "EmpID":  [1, 2, 3, 4, 5, 6],
    "Name":   ["Ahmed", "Fatima", "Bilal", "Ayesha", "Usman", "Zainab"],
    "DeptID": [10, 20, 10, 30, 20, 40],   # note: 40 has no matching department
    "Salary": [55000, 72000, 88000, 61000, 95000, 67000]
})

print("Employees DataFrame:")
print(employees)
print()
departments = pd.DataFrame({
    "DeptID":   [10, 20, 30, 50],          # note: 50 has no matching employee
    "DeptName": ["Engineering", "Marketing", "Design", "Finance"],
    "Location": ["Karachi", "Lahore", "Islamabad", "Peshawar"]
})

print("Departments DataFrame:")
print(departments)
print()


inner = pd.merge(employees, departments, on="DeptID", how="inner")
print("INNER JOIN (only matching DeptIDs):")
print(inner)
print()
left = pd.merge(employees, departments, on="DeptID", how="left")
print("LEFT JOIN (all employees, NaN where no match):")
print(left)
print()

employees_q1 = pd.DataFrame({
    "EmpID":  [1, 2, 3],
    "Name":   ["Ahmed", "Fatima", "Bilal"],
    "DeptID": [10, 20, 10],
    "Salary": [55000, 72000, 88000]
})

employees_q2 = pd.DataFrame({
    "EmpID":  [3, 4, 5],                    # EmpID 3 duplicates Q1
    "Name":   ["Bilal", "Ayesha", "Usman"],
    "DeptID": [10, 30, 20],
    "Salary": [88000, 61000, 95000]
})

print("Q1 employees:")
print(employees_q1)
print()
print("Q2 employees:")
print(employees_q2)
print()

# Concatenate (stack vertically)
combined = pd.concat([employees_q1, employees_q2], ignore_index=True)
print("After pd.concat() (ignore_index=True):")
print(combined)
print()

deduped = combined.drop_duplicates()
print("After drop_duplicates():")
print(deduped)
print()

deduped_by_id = combined.drop_duplicates(subset=["EmpID"])
print("After drop_duplicates(subset=['EmpID']):")
print(deduped_by_id)
```
